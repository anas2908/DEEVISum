```python
import h5py
import os

# Specify the path to the HDF5 file
hdf5_file_path = "C:\\Users\\Desktop\\iPTNet Code & Data\\Code & Data\\data\\datasets\\SMI_CVD.h5"

# Create an HDF5 file (or open an existing one) in write mode
with h5py.File(hdf5_file_path, "a") as hdf5_file:
    # Create 50 folders within the HDF5 file
    for i in range(1, 51):
        folder_name = f"video_{i}"
        
        # Check if the folder already exists
        if folder_name not in hdf5_file:
            hdf5_file.create_group(folder_name)
            print(f"Folder '{folder_name}' created.")
        else:
            print(f"Folder '{folder_name}' already exists.")
```

```python
import h5py
import numpy as np

# Specify the path to the HDF5 file
hdf5_file_path = "C:\\Users\\Desktop\\iPTNet Code & Data\\Code & Data\\data\\datasets\\SMI_CVD.h5"

# Function to create datasets within a group
def create_datasets(group):
    # Define the dataset names
    dataset_names = ['change_points', 'features', 'gtscore', 'gtsummary', 'n_frame_per_segment', 'n_frames', 'n_steps', 'picks', 'user_summary']

    # Create datasets within the group
    for dataset_name in dataset_names:
        # Check if the dataset already exists
        if dataset_name not in group:
            # Create a dummy dataset (replace this with your actual data)
            dummy_data = np.zeros((10, 10))  # Example: 10x10 matrix of zeros
            group.create_dataset(dataset_name, data=dummy_data)
            print(f"Dataset '{dataset_name}' created in group '{group.name}'.")
        else:
            print(f"Dataset '{dataset_name}' already exists in group '{group.name}'.")

# Open the HDF5 file in read-write mode
with h5py.File(hdf5_file_path, "a") as hdf5_file:
    # Iterate through each group in the HDF5 file
    for group_name, group in hdf5_file.items():
        # Check if the group contains datasets
        if isinstance(group, h5py.Group):
            # Create datasets within the group
            create_datasets(group)
```
